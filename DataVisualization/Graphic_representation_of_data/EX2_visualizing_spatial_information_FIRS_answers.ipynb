### Generate cartographic visualization with Vega-Altair

Vega allows for the creation of map-based visualizations using the same approach. Various types of symbols (marks) are available to create such visualizations, like:
- "geoshape" is used to represent the geometry of polygonal features stored in GeoJSON or TopoJSON format (mark_geoshape).
- "point" allows for the visualization of punctual geometries based on geographic coordinates stored in a table dataset.
- "rule" allows for the visualization of linear geometries based on the geographic coordinates of two points on a line, stored in a table dataset.
- "text" allows for the placement of labels, based on punctual coordinates.

Creating map visualizations requires specifying the projection used for projecting geographic coordinates on the plane map (project function).

A basic visualization with Vega can be the representation of the geographic features' geometry itself (contours and position). In order to visualize information related to the geographic features, this information has to 1. being integrated in the features' attribute table, or 2. being linked from an external table using a key-attribute. A geographical and a non-geographical datasets can be joined using the "lookup" property (LookupData function). Examine these various elements in the example here: https://altair-viz.github.io/gallery/choropleth.html .

In JSON specification of Vega and Vega-Lite libraries, different elements can be combined into a single visualization using the "layer" property. In Vega-Altair API approach, each component of the map can be stored in a variable, and combined later by "concatening" them using the "+" operator. Examine how these elements are combined here: https://altair-viz.github.io/gallery/london_tube.html .

You can find a example of map below. The join with table data is yet to be done.

```python
# Load the random GeoJSON data
altair.Chart(data_geojson_regions).mark_geoshape(stroke='black').encode(
    tooltip=[
                altair.Tooltip("properties.libelle_geog", type='nominal', title="name")#use properties node to access the attributes of geojson files
            ]
).properties(
    width=500,
    height=300
)
```

```python
datafile = r".\data\Arbre.csv" #locating Arbre.csv dataset on your computer
df = pandas.read_csv(datafile,sep=';') #importing Arbre.csv dataset into a Pandas dataframe
df_2015 = df.loc[df['CAMPAGNE'] == 2015] #creating another dataframe by extracting all data from Arbre.csv for the 2005 measurement campaign
df_2015['Vpondere'] = df_2015.V * df_2015.W
df_2015 = df_2015.dropna(subset=['Vpondere'])
df_2015['Vpondere'] = df_2015['Vpondere'].astype(int)
df_placette = pandas.read_csv(r".\data\Placette.csv",sep=';') #importing Arbre.csv dataset into a Pandas dataframe
df_placette_2015 = df_placette.loc[df_placette['CAMPAGNE'] == 2015] #creating another dataframe by extracting all data from Arbre.csv for the 2005 measurement campaign
V_by_spot = df_2015.groupby(['IDP'], sort=True)['Vpondere'].sum()
datafile_placette_2015_V = pandas.merge(df_placette_2015, V_by_spot, on='IDP', how='inner')
```

```python
volume_by_departements = datafile_placette_2015_V.groupby(['DEP'])["Vpondere"].sum().reset_index(name ='Volume')
volume_by_departements = volume_by_departements.dropna(subset=['DEP'])
volume_by_departements['DEP'] = volume_by_departements['DEP'].astype(str)
```

```python
volume_by_regions = datafile_placette_2015_V.groupby(['RAD13'])["Vpondere"].sum().reset_index(name ='Volume')
volume_by_regions = volume_by_regions.dropna(subset=['RAD13'])
volume_by_regions['RAD13'] = volume_by_regions['RAD13'].astype(str)
```

```python
volume_by_greco = datafile_placette_2015_V.groupby(['GRECO'])["Vpondere"].sum().reset_index(name ='Volume')
volume_by_greco = volume_by_greco.dropna(subset=['GRECO'])
volume_by_greco['GRECO'] = volume_by_greco['GRECO'].astype(str)
```

```python
volume_by_SER = datafile_placette_2015_V.groupby(['SER'])["Vpondere"].sum().reset_index(name ='Volume')
volume_by_SER = volume_by_SER.dropna(subset=['SER'])
volume_by_SER['SER'] = volume_by_SER['SER'].astype(str)
```

##### Question 2

Color is not the most efficient to directly display quantities, as it allows to represent order, but not proportions.
As consequence, choropleth maps displaying absolute numbers are considered as very bad in classical cartography.
Here, the proportionnal points allows directly to interpret that one region has "twice wood volume" than another. In order to make the same interpretation with the choropleth map, the user's eye has to go back and forth several times between the map and the legend.
Using color in maps is more accepted for density values (absolute values divided by population or surface), or for maps using a regular division of space (even if the better efficiency of size still goes on).


```python
simple_choroplethmap = altair.Chart(data_geojson_regions).mark_geoshape().transform_lookup(
    lookup='properties.code_geog',
    from_=altair.LookupData(volume_by_regions, 'RAD13', ['Volume'])
).encode(
    color='Volume:Q',
    tooltip=[
                altair.Tooltip("properties.libelle_geog", type='nominal', title="code"),
                altair.Tooltip("Volume", type='quantitative', title="wood volume")
            ]
)

background = altair.Chart(data_geojson_regions).mark_geoshape(fill='lightgray', stroke='white').encode(
    tooltip=[
                altair.Tooltip("properties.libelle_geog", type='nominal', title="code")
            ]
).project('mercator')

propcircle = altair.Chart(data_geojson_regions).mark_circle().transform_lookup(
    lookup='properties.code_geog',
    from_=altair.LookupData(volume_by_regions, 'RAD13', ['Volume'])).encode(
    longitude='properties.x:Q',
    latitude='properties.y:Q',
    size=altair.Size('Volume:Q'),
    color=altair.value('steelblue'),
    tooltip=[
                altair.Tooltip("properties.libelle_geog", type='nominal', title="code"),
                altair.Tooltip("Volume", type='quantitative', title="wood volume")
            ]
)
    
simple_choroplethmap | background + propcircle
```

##### Question 4

These different maps depict the Map Area Unit Problem (MAUP), which corresponds to the influence of the spatial grid used to discretize your initial data, on the aggregated data and its representation.

To tackle this problem, and not introduce a bias in the phenomena interpretation, the data can be represented either through regular grids (square or hexagons). However this regular sampling of space hides boundaries and spatial shapes users can be used to observe, and make spatial contextualization more difficult.
Another solution is the use of spatial division those geometry fits the spatial dynamic of the represented phenomena, like here with great ecoregions and sylvoecoregions.

```python
background_regions = altair.Chart(data_geojson_regions).mark_geoshape(stroke='black').transform_lookup(
    lookup='properties.code_geog',
    from_=altair.LookupData(volume_by_regions, 'RAD13', ['Volume'])).transform_calculate(
    dens='datum.Volume/datum.properties.area'
).encode(
    color='dens:Q',
    tooltip=[
                altair.Tooltip("properties.libelle_geog", type='nominal', title="code"),
                altair.Tooltip("dens", type='quantitative', title="wood dens")
            ]
).project('mercator')

propcircle_regions = altair.Chart(data_geojson_regions).mark_circle(fill='black').transform_lookup(
    lookup='properties.code_geog',
    from_=altair.LookupData(volume_by_regions, 'RAD13', ['Volume'])).encode(
    longitude='properties.x:Q',
    latitude='properties.y:Q',
    size=altair.Size('Volume:Q'),
    tooltip=[
                altair.Tooltip("properties.libelle_geog", type='nominal', title="code"),
                altair.Tooltip("Volume", type='quantitative', title="wood volume")
            ]
)
  
background_grandesecoregions = altair.Chart(data_geojson_grandesecoregions).mark_geoshape(stroke='black').transform_lookup(
    lookup='properties.code_geog',
    from_=altair.LookupData(volume_by_greco, 'GRECO', ['Volume'])).transform_calculate(
    dens='datum.Volume/datum.properties.area'
).encode(
    color='dens:Q',
    tooltip=[
                altair.Tooltip("properties.libelle_geog", type='nominal', title="code"),
                altair.Tooltip("dens", type='quantitative', title="wood dens")
            ]
).project('mercator')

propcircle_grandesecoregions = altair.Chart(data_geojson_grandesecoregions).mark_circle(fill='black').transform_lookup(
    lookup='properties.code_geog',
    from_=altair.LookupData(volume_by_greco, 'GRECO', ['Volume'])).encode(
    longitude='properties.x:Q',
    latitude='properties.y:Q',
    size=altair.Size('Volume:Q'),
    tooltip=[
                altair.Tooltip("properties.libelle_geog", type='nominal', title="code"),
                altair.Tooltip("Volume", type='quantitative', title="wood volume")
            ]
)
    
background_regions + propcircle_regions | background_grandesecoregions + propcircle_grandesecoregions 
    
```

##### Question 2

Volume change can be positive or negative.
We can use size to represent the absolute value of change, combine with color using a bivariate color to represent if the change is positive or negative. Or using a simple green/red distinction.

```python
volume_by_GRECO_and_date_2015 = volume_by_GRECO_and_date[volume_by_GRECO_and_date["CAMPAGNE"] == 2015]
volume_by_GRECO_and_date_2015 = volume_by_GRECO_and_date_2015[['GRECO','Volume']]
volume_by_GRECO_and_date_2021 = volume_by_GRECO_and_date[volume_by_GRECO_and_date["CAMPAGNE"] == 2021]
volume_by_GRECO_and_date_2021 = volume_by_GRECO_and_date_2021[['GRECO','Volume']]
volume_change_by_GRECO = pandas.merge(volume_by_GRECO_and_date_2015, volume_by_GRECO_and_date_2021, on=['GRECO'], how='inner')
volume_change_by_GRECO['Volume_change'] = volume_change_by_GRECO['Volume_y'] - volume_change_by_GRECO['Volume_x']
volume_change_by_GRECO['abs_Volume_change'] = abs(volume_change_by_GRECO['Volume_change'])
```

```python
background_grandesecoregions = altair.Chart(data_geojson_grandesecoregions).mark_geoshape(fill='grey', stroke='white').encode(
    tooltip=[
                altair.Tooltip("properties.libelle_geog", type='nominal', title="code")
            ]
).project('mercator')

propcircle_grandesecoregions = altair.Chart(data_geojson_grandesecoregions).mark_circle().transform_lookup(
    lookup='properties.code_geog',
    from_=altair.LookupData(volume_change_by_GRECO, 'GRECO', ['Volume_change','abs_Volume_change'])).encode(
    longitude='properties.x:Q',
    latitude='properties.y:Q',
    size=altair.Size('abs_Volume_change:Q'),
    color=altair.Color('Volume_change:Q',scale=altair.Scale(domainMid=0, range='diverging')),
    tooltip=[
                altair.Tooltip("properties.libelle_geog", type='nominal', title="code"),
                altair.Tooltip("Volume_change", type='quantitative', title="wood volume change")
            ]
)

background_grandesecoregions + propcircle_grandesecoregions
```
