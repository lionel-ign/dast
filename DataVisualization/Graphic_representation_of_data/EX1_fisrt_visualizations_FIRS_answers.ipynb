### Create visualizations with Vega-Altair

#### What is Vega-Altair?

We will mainly use here the [Vega-Altair](https://altair-viz.github.io/) module to create visualizations. Altair is a python module adapted from the [Vega](https://vega.github.io/vega/) javascript library. Vega (and its lighter version Vega-Lite) is a framework based on Leland Wilkinson's Grammar of Graphics for creating SVG visualizations using [D3](https://d3js.org/) JavaScript library. Vega offers a declarative language for creating visualizations; they use a visualization grammar structured in JSON format, where we define the graphical layout (the <b>mark</b>) of the symbols representing our data, and the visual channels used for representing the information (the <b>encoding</b>). Based on these definitions — which specify which data components to visualize and how — the corresponding visualization is generated in SVG format using D3 (other outputs can be specified: png, canvas,...). Vega-Altair follows the same logic, being a declarative visualization API-based enabling the creation of simple visualizations without the need for complex coding. You can find various examples [here](https://altair-viz.github.io/gallery/index.html#example-gallery). The first step is to install Altair and import it on your notebook.

#### How it works?

In the following code lines, we define the parameters of a visualization representing the distribution according to one variable.

We specify a type of [mark](https://altair-viz.github.io/user_guide/marks/index.html), which define which kind of graphic object will be used for our visualization's symbols.

We specify an [encoding](https://altair-viz.github.io/user_guide/encodings/index.html), which define which symbols' visual channel will be used to represent the information, and how it will be represented. When defining an encoding, we must specify what information or constant value that encoding represents. This can be a dataset attribute, a constant graphical value, a specific value linked to the dataset. For each attribute variable defined in the encoding, we specify the type of variable being represented. There are 5 types: quantitative, ordinal, nominal (categorical), temporal, and GeoJSON for geographic data in GeoJSON format.
Depending on the type selected to describe the attribute, the chosen visual variable will be used differently to encode the information; for example, the color scale applied will differ depending on whether an attribute is specified as categorical or ordinal.

```python
altair.Chart(df_2015).mark_tick(size=60).encode(
    altair.X('V', type='quantitative')
).properties(
    width=1000
)
```

#### Vega or another thing?

All Vega components are a good introduction materials for visualization conception, since it is focused on the declaration of visualization specifications. 
However, some complex visualizations are difficult to be created with Vega. For such kind of visualization, a more complex solution is necessary ([D3](https://d3js.org/), [Matplotlib](https://matplotlib.org/), [Plotly](https://plotly.com/python/), etc).  

Vega-Altair is only able to deal with not-to-massive datasets (up to 100 000 individuals). If you have to deal with massive datasets, other modules such as Matplotlib or Plotly are necessary.

##### Question 7

```python
bins = 50
df_eco_2015_V['diff_prof'] = df_eco_2015_V.PROF2 - df_eco_2015_V.PROF1
points = altair.Chart(df_eco_2015_V).mark_circle(filled=False, color='red').encode(
    altair.X('PENT2', type='quantitative').bin(maxbins=bins),
    altair.Y('Vpondere', type='quantitative').bin(maxbins=bins),
    altair.Size("count()")
).properties(
    width=500,
    height=500
)

heatmap = altair.Chart(df_eco_2015_V).mark_rect().encode(
    altair.X('PENT2', type='quantitative').bin(maxbins=bins),
    altair.Y('Vpondere', type='quantitative').bin(maxbins=bins),
    altair.Color("mean(diff_prof)").scale(scheme="greenblue"),
    tooltip=[
        altair.Tooltip("PENT2", title="mean slope"),
        altair.Tooltip("Vpondere", title="wood volume"),
        altair.Tooltip("count()", title="number of sites")
    ]
).properties(
    width=500,
    height=500
)

heatmap + points
```

The line chart allows a better interpretation of continuity between different timesteps. 
The bar chart allows a better comparison between timesteps, but can be more difficult to read than the linechart in case of a lot of timesteps.

The Y-axis should show the distance between the axis bottom and the 0 value, using a dotted axis segment for example.

##### Question 3

The multi-line chart allows a better comparison between different individuals.
Stacked charts allow a better interpretation of the overall evolution of the dataset, with cumulative evolution.

```python
multi_line_chart = altair.Chart(df_by_date_and_type).mark_line().encode(
    x='CAMPAGNE',
    y=altair.Y('TotVolume', scale=altair.Scale(domain=[250000, 400000])),
    color='Type_a'
).properties(
    width=300
)

stacked_bart_chart = altair.Chart(df_by_date_and_type).mark_bar(clip=True).encode(
    altair.X("CAMPAGNE", type='ordinal'),
    y=altair.Y('TotVolume', scale=altair.Scale(domain=[0, 800000])),
    color='Type_a'
).properties(
    width=300
)

stacked_area_chart = altair.Chart(df_by_date_and_type).mark_area(clip=True).encode(
    altair.X("CAMPAGNE", type='ordinal'),
    y=altair.Y('TotVolume', scale=altair.Scale(domain=[0, 800000])),
    color='Type_a'
).properties(
    width=300
)

multi_line_chart | stacked_bart_chart | stacked_area_chart
```
