```python
import pandas #importing pandas
import geopandas #importing geopandas
import altair #importing altair
altair.data_transformers.enable("vegafusion") # this command allows vega-altair to deal with relatively massive data (up to 100 000 individuals)
```

```python
#importing departements geosjson
url_departements = r".\data\departements_wgs84.geojson"
geojson_file_departements = geopandas.read_file(url_departements)
geojson_file_departements = geojson_file_departements.dropna(subset=['code_geog'])
geojson_file_departements['code_geog'] = geojson_file_departements['code_geog'].astype(str)
data_geojson_departements = altair.InlineData(values=geojson_file_departements.to_json(), format=altair.DataFormat(property='features',type='json')) 
```

```python
# importing regions geojson
url_regions = r".\data\regions_wgs84.geojson"
geojson_file_regions = geopandas.read_file(url_regions)
geojson_file_regions = geojson_file_regions.dropna(subset=['code_geog'])
geojson_file_regions['code_geog'] = geojson_file_regions['code_geog'].astype(str)
data_geojson_regions = altair.InlineData(values=geojson_file_regions.to_json(), format=altair.DataFormat(property='features',type='json')) 
```

```python
#importing sylvoecoregions geosjson
url_sylvoecoregions = r".\data\sylvoecoregions_wgs84.geojson"
geojson_file_sylvoecoregions = geopandas.read_file(url_sylvoecoregions)
geojson_file_sylvoecoregions = geojson_file_sylvoecoregions.dropna(subset=['code_geog'])
geojson_file_sylvoecoregions['code_geog'] = geojson_file_sylvoecoregions['code_geog'].astype(str)
data_geojson_sylvoecoregions = altair.InlineData(values=geojson_file_sylvoecoregions.to_json(), format=altair.DataFormat(property='features',type='json')) 
```

```python
#importing great eco regions geosjson
url_grandesecoregions = r".\data\grandesecoregions_wgs84.geojson"
geojson_file_grandesecoregions = geopandas.read_file(url_grandesecoregions)
geojson_file_grandesecoregions = geojson_file_grandesecoregions.dropna(subset=['code_geog'])
geojson_file_grandesecoregions['code_geog'] = geojson_file_grandesecoregions['code_geog'].astype(str)
data_geojson_grandesecoregions = altair.InlineData(values=geojson_file_grandesecoregions.to_json(), format=altair.DataFormat(property='features',type='json')) 
```

### Generate cartographic visualization with Vega-Altair

Vega allows for the creation of map-based visualizations using the same approach. Various types of symbols (marks) are available to create such visualizations, like:
- "geoshape" is used to represent the geometry of polygonal features stored in GeoJSON or TopoJSON format (mark_geoshape).
- "point" allows for the visualization of punctual geometries based on geographic coordinates stored in a table dataset.
- "rule" allows for the visualization of linear geometries based on the geographic coordinates of two points on a line, stored in a table dataset.
- "text" allows for the placement of labels, based on punctual coordinates.

Creating map visualizations requires specifying the projection used for projecting geographic coordinates on the plane map (project function).

A basic visualization with Vega can be the representation of the geographic features' geometry itself (contours and position). In order to visualize information related to the geographic features, this information has to 1. being integrated in the features' attribute table, or 2. being linked from an external table using a key-attribute. A geographical and a non-geographical datasets can be joined using the "lookup" property (LookupData function). Examine these various elements in the example here: https://altair-viz.github.io/gallery/choropleth.html .

In JSON specification of Vega and Vega-Lite libraries, different elements can be combined into a single visualization using the "layer" property. In Vega-Altair API approach, each component of the map can be stored in a variable, and combined later by "concatening" them using the "+" operator. Examine how these elements are combined here: https://altair-viz.github.io/gallery/london_tube.html .

You can find a example of map below. The join with table data is yet to be done.

```python
# Load the random GeoJSON data
altair.Chart(data_geojson_regions).mark_geoshape(stroke='black').encode(
    tooltip=[
                altair.Tooltip("properties.libelle_geog", type='nominal', title="name")#use properties node to access the attributes of geojson files
            ]
).properties(
    width=500,
    height=300
)
```
